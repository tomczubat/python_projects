```python
import pandas as pd
```

```python
df = pd.read_csv('customers.csv')
```

```python
display(df)
```

```python
display(len(df))
```

```python
df.dropna()
df.fillna(0)
```

```python
df = df[df['customer_id'] < 4]
display(df)
```

```python
# How do you remove duplicate rows?
df.drop_duplicates()

# What is the use of apply()?
df['Salary'] = df['Salary'].apply(lambda x: x*1.1)

# How can you rename columns?
df.rename(columns={'old_name':'new_name'})

# How do you sort a DataFrame?
df.sort_values(by='Salary', ascending=False)

# What is the difference between concat() and merge()?
Answer:
- concat() stacks DataFrames.
- merge() joins DataFrames using keys.

# How do you find unique values?
df['Department'].unique()
```
